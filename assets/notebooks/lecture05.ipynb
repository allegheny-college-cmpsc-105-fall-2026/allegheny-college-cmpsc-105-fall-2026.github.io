## Storing Strings in Variables

A string is a sequence of characters. In Python, we can store strings in variables just like we do with numbers. Strings are enclosed in either single quotes (`'`) or double quotes (`"`).

```python
my_string = "Hello, World!"
```

```python
# example operations on strings
print(my_string.upper())  # Convert to uppercase
print(my_string.lower())  # Convert to lowercase
print(my_string[:5])  # Slice the string to get "Hello"
```

## Printing Output

f-strings allow you to embed expressions inside strings, using curly braces `{}`.


```python
print(f'The value of x is {x} and the value of y is {y}.')  # Using f-strings to embed variables in strings
```

## Logical Statements

Logical statements allow us to make decisions in our code and evaluate to either `True` or `False`. We can use logical operators such as `and`, `or`, and `not` to combine multiple conditions.

```python
2 > 1
```

```python
print(1 >= 2) # 1 is greater than or equal to 2? False
print(1 < 2)  # 1 is less than 2? True
print(x == 10) # x is equal to 10? True
print(my_string != "Hello") # my_string is not equal to "Hello"? True
```

Helpful operations: 
- `==` : equal to
- `!=` : not equal to
- `>` : greater than
- `<` : less than
- `>=` : greater than or equal to
- `<=` : less than or equal to

We can also combine logical statements using `and`, `or`, and `not`. 
- `and` : returns `True` if both statements are true
- `or` : returns `True` if at least one statement is true
- `not` : returns `True` if the statement is false

```python
# we can filter rows using a logical statement
# using brackets this way is called indexing

# For example, we can filter the DataFrame to only include rows where the 'week' column is equal to 3.
df_week3 = df[df['week'] == 3] 
df_week3
```

```python
# we can also filter columns using a list of column names. For example, we can select only the 'week' and 'coffee_consumed' columns from the DataFrame.
df_week_coffee = df[['week', 'coffee_consumed']]
df_week_coffee
```
