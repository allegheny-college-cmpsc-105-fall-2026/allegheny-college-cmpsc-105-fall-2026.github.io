```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Set the visual theme for the notebook
sns.set_theme(style="whitegrid")

# Load built-in datasets for the lecture
mpg = sns.load_dataset("mpg")
tips = sns.load_dataset("tips")
```

## Univariate Plots

### Quantitative - Histogram
The histogram is a graphical representation of the distribution of a dataset. It is created by dividing the range of the data into intervals, known as bins, and counting the number of observations that fall into each bin. The height of each bar in the histogram represents the frequency or count of observations in that bin.

**Key Parameter:** bins - The number of intervals (bins) to divide the data into. A higher number of bins can provide a more detailed view of the distribution, while a lower number of bins can give a more general overview. Can be difficult to choose the right number of bins. One heuristic is to use the square root of the number of observations in the dataset as the number of bins.

```python
sns.histplot(data=mpg, x='mpg', bins=20)
plt.title('Distribution of Miles Per Gallon')
plt.show()
```

### Qualitative - Countplot
The countplot is a bar plot that shows the counts of observations in each categorical bin using bars. It is useful for visualizing the distribution of categorical variables.

```python
sns.countplot(data=tips, x='day')
plt.title('Number of Tables Served per Day')
plt.show()
```

## Multivariate Plots

### Quant-Quant - Scatterplot
The scatterplot is a graphical representation of the relationship between two quantitative variables. Each point on the plot represents an observation in the dataset, with its position determined by the values of the two variables.

**Key Parameters:**
- **hue** - A variable that defines the color of the points in the scatterplot. 
- **size** - The size of the points in the scatterplot.
- **style** - The style of the points in the scatterplot. 
- **alpha** - The transparency level of the points in the scatterplot. A lower alpha value can help to reduce overplotting when there are many points in the plot, making it easier to see the overall distribution of the data.

```python
sns.scatterplot(data=mpg, x='weight', y='mpg', alpha=0.7)
plt.title('Vehicle Weight vs. MPG')
plt.show()
```

### Quant-Quant - Line Plot
The line plot is a type of chart used to visualize the relationship between two quantitative variables. It is created by plotting points on a graph and connecting them with lines. Line plots are particularly useful for visualizing trends over time or ordered categories.

**Key Parameters:**
- **hue** - A variable that defines the color of the lines in the line plot. 
- **style** - The style of the lines in the line plot.

```python
sns.lineplot(data=tips, x='size', y='total_bill')
plt.title('Party Size vs. Total Bill')
plt.show()
```

### Quant-Qual - Boxplot
The boxplot is a standardized way of displaying the distribution of data based on a five-number summary: minimum, first quartile (Q1), median, third quartile (Q3), and maximum. It is particularly useful for identifying outliers and understanding the spread and skewness of the data.

```python
sns.boxplot(data=tips, x='day', y='total_bill')
plt.title('Total Bill Distributions by Day')
plt.show()
```

### Quant-Qual - Violin Plot
The violin plot is a method of plotting numeric data and can be understood as a combination of a boxplot and a kernel density plot. It provides a deeper understanding of the distribution of the data, including its probability density, and is particularly useful for comparing distributions between multiple groups or categories.

```python
sns.violinplot(data=tips, x='day', y='total_bill')
plt.title('Total Bill Distributions by Day')
plt.show()
```

### Qual-Qual - Two-way Countplot
The two-way countplot is a bar plot that shows the counts of observations in each combination of two categorical variables. It is useful for visualizing the relationship between two categorical variables and can help identify patterns or associations between them.

```python
sns.countplot(data=tips, x='day', hue='smoker', palette='magma')
plt.title('Smoker Status by Day')
plt.show()
```

### Qual-Qual - Heatmap
The heatmap is a graphical representation of data where individual values are represented as colors. It is particularly useful for visualizing the relationship between two categorical variables, especially when the data is in a matrix format. The intensity of the color indicates the magnitude of the value, allowing for quick identification of patterns and correlations between the variables.

```python
crosstab = pd.crosstab(tips['day'], tips['smoker'])
sns.heatmap(crosstab, annot=True, cmap='Blues', fmt='d')
plt.title('Heatmap of Smoker Status by Day')
plt.show()
```
